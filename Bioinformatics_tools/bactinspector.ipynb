# **BactInspectorMax – Reference Genome Identification**

## **Tool Information**

- **Tool:** BactInspectorMax (command-line tool: `bactinspector`)    
- **Input:** Genome assemblies in FASTA format generated from the Nextflow pipeline  
- **Organism:** *Acinetobacter baumannii*  
- **Analysis type:** Mash-based genome distance estimation  

BactInspectorMax is designed to identify the most appropriate reference genomes for a given set of bacterial isolates by comparing genomic similarity and relatedness. It enables the selection of closely related reference sequences that can be used for downstream analyses such as comparative genomics and alignment-based studies. 

The tool takes genome assemblies in FASTA format as input and provides ranked reference genomes based on similarity metrics. In this study, BactInspectorMax is used to ensure that downstream analyses are performed using biologically relevant and closely related reference genomes, thereby improving the accuracy of comparative and contextual interpretations.

## **BactInspectorMax Environment Setup and Installation**

A dedicated Conda environment was created for BactInspectorMax to ensure reproducible execution and avoid dependency conflicts. The environment included all required Python libraries and the bundled RefSeq Mash sketches used for identifying the closest reference genomes.

```python
%%bash
# Enable conda inside Jupyter
source /home/anaconda/miniconda3/etc/profile.d/conda.sh

# Create a clean environment for BactInspectorMax
conda create -n bactinspectormax_aba python=3.9 -y

# Activate environment
conda activate bactinspectormax_aba

# Install the packaged, stable version with bundled RefSeq data
pip install bactinspectorMax

# Verify Installation
bactinspector --version
bactinspector --help
```

# **Input Files**

The analysis requires assembled bacterial genomes in FASTA format as input.  
Each genome assembly should contain high-quality contigs suitable for comparative genomic analysis.

# **Run BactInspectorMax – Closest Reference Genome**

BactInspectorMax identifies the closest RefSeq reference genome based on Mash distance. The analysis was performed in parallel using 12 threads.

```python
%%bash
# Enable conda inside Jupyter
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate bactinspectormax_aba

# Define paths
INPUT_DIR=/data/internship_data/nidhi/aba/new_output/nextflow_output/assemblies
OUTPUT_DIR=/data/internship_data/nidhi/aba/new_output/bactinspector_output

# Create output directory
mkdir -p "$OUTPUT_DIR"

# Run closest_match using bundled RefSeq Mash sketches
bactinspector closest_match \
    -i "$INPUT_DIR" \
    -o "$OUTPUT_DIR" \
    -f "*.fasta" \
    -p 12 \
    -mp /usr/bin
```

# **Output Files**

BactInspectorMax generates Mash sketch files and summary tables containing the closest reference genome matches identified during analysis.  
These outputs are used to determine genetically related reference genomes for downstream comparative and phylogenetic analyses.

| Output File | Description |
|---|---|
| `closest_matches_*.tsv` | Tab-separated summary of the closest reference genome matches and similarity metrics |
| `*.short.msh` | Mash sketch files generated for rapid genomic distance estimation |
