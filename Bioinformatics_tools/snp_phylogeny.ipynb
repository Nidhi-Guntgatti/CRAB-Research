# **SNP-based Phylogenetic Analysis using GHRU SNP Phylogeny Pipeline**

This notebook documents reference-based SNP phylogeny analysis of *Acinetobacter baumannii* isolates using the GHRU SNP phylogeny Nextflow workflow.

## **Tool Information**

- Tool: GHRU SNP Phylogeny Pipeline

  Available at:
  https://gitlab.com/cgps/ghru/pipelines/snp_phylogeny
- Workflow manager: Nextflow
- Organism: Acinetobacter baumannii
- Analysis type: Reference-based SNP phylogeny

The GHRU SNP phylogeny pipeline performs read mapping against a reference genome, SNP calling, core SNP extraction, and phylogenetic tree generation.

The workflow internally uses multiple bioinformatics tools including:

- BWA-MEM for read alignment
- samtools and bcftools for variant calling
- IQ-TREE for phylogenetic tree construction

The pipeline was executed using the Nextflow workflow framework.

## **Conda Environment Creation**

A dedicated conda environment was created for executing the GHRU SNP phylogeny pipeline and its dependencies.

```python
%%bash

conda create -n snp_phylogeny -y
```

Nextflow was installed as the workflow management system for execution of the GHRU SNP phylogeny pipeline.

The installation script automatically downloaded the Nextflow executable and its required dependencies.

## **Input Files**

The workflow requires paired-end Illumina FASTQ files for each isolate and a reference genome assembly in FASTA format.

Input files include:

- Forward reads (R1 FASTQ)
- Reverse reads (R2 FASTQ)
- Reference genome FASTA

Each row in the sample sheet corresponds to one isolate.

## **Reference Genome**

A complete *Acinetobacter baumannii* reference genome assembly obtained from
the NCBI RefSeq database was used for read mapping and SNP identification.

Reference genome details:

- Assembly accession: GCF_000828935.1
- Source database: NCBI RefSeq
- File used: `chromosome.fasta`

All sequencing reads were aligned against the same reference genome to ensure
consistent SNP calling and phylogenetic inference across isolates.

## **Execution Strategy**

The GHRU SNP phylogeny pipeline was executed using Nextflow. The workflow
processed all isolates in parallel and performed:

- Read mapping
- Variant calling
- Core SNP extraction
- Phylogenetic inference

Nextflow automatically handled job parallelisation, dependency management, and
workflow reproducibility.

```python
%%bash

source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate snp_phylogeny

export NXF_VER=22.10.8

cd /data/internship_data/nidhi/aba/snp_phylogeny_run/snp_phylogeny

/home/nidhi/jupyter_notebooks/tools/nextflow run snp_phylogeny.nf \
    --input_dir /data/internship_data/nidhi/aba/fastqs/set_1 \
    --fastq_pattern "*_{1,2}.fastq.gz" \
    --reference /data/internship_data/nidhi/aba/output/refseq_ncbi/ncbi_dataset_2/data/GCF_000828935.1/chromosome.fasta \
    --output_dir /data/internship_data/nidhi/aba/output/snp_phylogeny/ref_2 \
    --tree \
    --max_cpus 32 \
    --max_memory 64.GB \
    -resume
```

## **Output Files**

The SNP phylogeny workflow generates multiple intermediate and final outputs.
Important files used for downstream analysis include:

- `aligned_pseudogenome.fas`  
  Core genome pseudoalignment generated after reference-based SNP calling.

- `aligned_pseudogenome.variants_only.fas`  
  Alignment containing only polymorphic SNP sites across all isolates.

- `aligned_pseudogenome.variants_only.tre`  
  Phylogenetic tree inferred from the SNP-only alignment.

- `*.filtered.bcf`  
  Binary variant call files containing filtered SNP variants identified
  relative to the reference genome.

- `low_quality_pseudogenomes.tsv`  
  Summary table of isolates identified as low-quality during pseudoalignment
  generation.
