# **Genome Annotation – Bakta**

## **Tool Information**

- **Tool:** Bakta v1.11.4
- **Input:** Assembled genome FASTA files  
- **Organism:** *Acinetobacter baumannii*  
- **Analysis type:** Genome annotation and functional characterization  

Bakta is a standardized bacterial genome annotation tool designed to generate high-quality annotations from assembled genomes while reducing inconsistencies associated with fragmented or misannotated genes.

It performs automated identification and annotation of coding sequences, RNAs, and other genomic features using curated protein databases and HMM-based searches, producing standardized outputs compatible with downstream comparative genomic analyses.

This notebook documents the annotation of *Acinetobacter baumannii* genome assemblies and generation of annotation files for downstream gene context, synteny, and comparative genomic analyses.

## **Execute Bakta Annotation**

Bakta genome annotation was executed on all assembled genomes using the downloaded reference database. The analysis generated standardized annotation outputs including GFF3, GBFF, FAA, FFN and annotation summary files for downstream comparative genomic analyses.

`export MPLBACKEND=Agg` was used to set the Matplotlib backend to a non-interactive rendering mode compatible with command-line and Jupyter-based execution environments. This prevents backend-related plotting errors during Bakta execution.

```python
%%bash

source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate bakta

export MPLBACKEND=Agg

INPUT_DIR="/data/internship_data/nidhi/aba/new_output/nextflow_output/assemblies"
OUTPUT_DIR="/data/internship_data/nidhi/aba/new_output/bakta_output"
DB_PATH="/data/internship_data/nidhi/aba/output/bakta_output/database/db"

mkdir -p ${OUTPUT_DIR}

for genome in ${INPUT_DIR}/*.short.fasta; do
    sample=$(basename ${genome} .short.fasta)

    bakta \
        --db ${DB_PATH} \
        --output ${OUTPUT_DIR}/${sample} \
        --prefix ${sample} \
        --threads 32 \
        ${genome}

done
```

## **Expected Output Files**

Bakta generates multiple annotation outputs containing predicted genes,
functional annotations, and genome feature summaries. Important output files
include:

- `.gff3`  
  Genome annotation file containing genomic feature coordinates and
  annotations in GFF3 format.

- `.gbff`  
  GenBank-formatted annotation file compatible with downstream genome
  browsers and comparative genomics tools.

- `.faa`  
  Amino acid sequences of predicted protein-coding genes.

- `.ffn`  
  Nucleotide sequences of predicted coding sequences (CDS).

- `.fna`  
  Nucleotide FASTA file of the annotated genome assembly.

- `.embl`  
  EMBL-formatted genome annotation file.

- `.tsv`  
  Tabular annotation summary containing predicted gene functions and feature
  information.

- `.json`  
  Structured JSON-formatted annotation output for downstream computational
  processing.

- `.log`  
  Workflow execution log generated during annotation.

- `.txt`  
  Annotation statistics and summary report.

- `.svg` and `.png`  
  Circular graphical genome visualizations generated by Bakta.

- `hypotheticals.faa`  
  Amino acid sequences of proteins annotated as hypothetical proteins.

- `hypotheticals.tsv`  
  Tabular summary of hypothetical protein annotations.

- `inference.tsv`  
  Annotation evidence and inference summary for predicted genomic features.
