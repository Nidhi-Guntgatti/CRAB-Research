# **KL and OC Locus Typing – Kaptive**

## **Tool Information**

- **Tool:** Kaptive v3.1.0
- **Input:** Assembled genome contigs (FASTA)
- **Organism:** *Acinetobacter baumannii*
- **Analysis type:** Capsule (KL) and lipooligosaccharide (OC) locus typing
- **Databases used:** ab_k (KL), ab_o (OC)

This notebook documents the identification of capsule (KL) and outer core lipooligosaccharide (OC) locus types in *Acinetobacter baumannii* genomes using Kaptive.

Kaptive performs locus typing by comparing genome assemblies against curated reference databases of known KL (capsule) and OC (outer core) biosynthesis loci. These loci are important epidemiological markers and are frequently associated with clonal lineages and virulence traits.

The analysis was performed on assembled genomes generated from the GHRU assembly pipeline.

## **Kaptive Execution**

Each genome assembly was processed in batch mode using a multi-threaded execution strategy.

The analysis was performed within a Conda-managed environment to ensure reproducibility and dependency control. The KL (ab_k) and OC (ab_o) reference databases were used for locus identification.

To ensure compatibility within the Jupyter environment, the matplotlib backend was configured to use a non-interactive setting. Runtime messages were allowed to execute normally, while output results were written directly to structured tab-separated result files for downstream analysis.

````python
%%bash

# Activate conda
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate kaptive

# Fix matplotlib backend issue (for jupyter incompatibility)
export MPLBACKEND=Agg

# Input and output paths
ASSEMBLY_DIR="/data/internship_data/nidhi/aba/new_output/nextflow_output/assemblies"
OUT_DIR="/data/internship_data/nidhi/aba/new_output/kaptive_output"
THREADS=16        # change if needed: 8, 16, 32

# Step 1: KL typing (capsule locus)
echo ">>> Running KL (capsule) typing..."

kaptive assembly \
    ab_k \
    ${ASSEMBLY_DIR}/*.fasta \
    -o ${OUT_DIR}/KL_results.tsv \
    --threads ${THREADS}

echo ">>> KL typing complete! Results saved to:"
echo "${OUT_DIR}/KL_results.tsv"

# Step 2: OC typing (lipooligosaccharide locus)
echo ">>> Running OC (lipooligosaccharide) typing..."

kaptive assembly \
    ab_o \
    ${ASSEMBLY_DIR}/*.fasta \
    -o ${OUT_DIR}/OC_results.tsv \
    --threads ${THREADS}

echo ">>> OC typing complete! Results saved to:"
echo "${OUT_DIR}/OC_results.tsv"
````

## **Kaptive Output**

Kaptive generated tab-separated result files summarizing KL and OC locus
assignments for each genome assembly.

Two primary output files were produced:

- KL_results.tsv → Capsule locus typing results
- OC_results.tsv → Outer core locus typing results

Each output file contains one row per isolate, including:

- Assigned locus type
- Match confidence level
- Percentage coverage
- Percentage identity
- Length discrepancies relative to reference loci
- Missing or extra gene counts

These results provide standardized KL and OC locus assignments and were used
for downstream epidemiological comparison alongside MLST, AMR gene profiling,
and plasmid typing analyses.
