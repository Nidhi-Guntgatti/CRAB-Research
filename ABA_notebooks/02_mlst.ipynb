# **Multilocus Sequence Typing (MLST)**

## **Tool Information**

- **Analysis type:** MLST (v2.23.0)
- **Input data:** Assembled genomes (FASTA)
- **Organism:** *Acinetobacter baumannii*
- **Typing scheme:** Oxford MLST

Following genome assembly, **multilocus sequence typing (MLST)** was performed to assign sequence types (STs) to each isolate. MLST assigns isolates to sequence types based on the allelic combinations of housekeeping genes defined in a species-specific typing scheme. This allows comparison of isolates across studies and supports population structure analysis.

For *Acinetobacter baumannii*, the Oxford MLST scheme was used.

MLST was performed using **mlst (2.23.0)** (Torsten Seemann), a command-line tool that assigns sequence types by querying assembled genome contigs against curated PubMLST schemes.

## **Installation**

### 1) Create a dedicated environment
Created a dedicated conda environment to isolate dependencies and avoid conflicts with other tools. This ensures a clean and reproducible setup for running MLST analyses.

```python
%%bash

conda create -n mlst_aba -y
```

### 2) Install MLST using mamba
Installed MLST using mamba, a faster dependency resolver than conda, to ensure efficient installation. MLST is installed from Bioconda along with required dependencies from conda-forge.

```python
%%bash

# Activate environment
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate mlst_aba

conda install -n base -c conda-forge mamba -y
mamba install -c bioconda -c conda-forge mlst -y
```

### 3) Verify Installation
Verify that MLST has been installed correctly by checking its version. This step confirms that the tool is ready for downstream sequence typing analysis.

```python
%%bash

# initialise conda
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate mlst_aba

# check mlst
mlst --version
```

### MLST Database

MLST automatically downloads scheme databases on first use.  

⚠️ Ensure internet access is available during the first run.

Alternatively, databases can be pre-downloaded and configured manually if working in a restricted environment.

### 3) Database Installation Manually (Optional)

```python
%%bash

# Create a directory for MLST databases
mkdir -p ~/mlst_db

# Set environment variable
export MLST_DB=~/mlst_db

# Download MLST Database
mlst --update

# Define path so mlst can always find the database
echo 'export MLST_DB=~/mlst_db' >> ~/.bashrc
source ~/.bashrc
```

## **MLST Execution**

The analysis was performed in a Conda-managed environment, standard output was written to a results file and execution logs were retained for reproducibility.

```python
%%bash

# initialise conda
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate mlst_aba

# define paths
ASSEMBLY_DIR=/data/internship_data/nidhi/aba/new_output/nextflow_output/assemblies
MLST_OUTDIR=/data/internship_data/nidhi/aba/new_output/mlst_output
LOGDIR=/data/internship_data/nidhi/aba/new_output/logs

# create required directories
mkdir -p $MLST_OUTDIR $LOGDIR

# run MLST with logging
mlst $ASSEMBLY_DIR/*.short.fasta \
> $MLST_OUTDIR/mlst_results.tsv \
2> $LOGDIR/mlst.log
```

## **MLST Output**

The MLST tool generates a primary output containing sequence type (ST) and allelic profiles for each isolate.
